
## 7. Machine Learning Challenge

Let’s try a **classification task**: Predict if a patient's test result will be **Normal** or **Abnormal**.

👉 **Tasks:**  
1. Select relevant features: Age, Gender, Medical Condition, Admission Type, Billing Amount.  
2. Encode categorical variables (e.g., using `pd.get_dummies`).  
3. Split the data into training and testing sets.  
4. Train **two models**:  
   - Decision Tree Classifier  
   - Random Forest Classifier  
5. Evaluate both models using:  
   - Accuracy score  
   - Confusion matrix  
6. Compare the results: Which model performed better? Why do you think so?  


```python

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay

# 1. Focus only on Normal vs Abnormal for simplicity
df_binary = df[df["Test Results"].isin(["Normal", "Abnormal"])]

X = df_binary[["Age", "Gender", "Medical Condition", "Admission Type", "Billing Amount"]]
y = df_binary["Test Results"]

# 2. Encode categorical variables
X = pd.get_dummies(X, drop_first=True)

# 3. Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

# 4. Train Decision Tree
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)

# 5. Train Random Forest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)

# 6. Evaluate both models
print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt))
print("Random Forest Accuracy:", accuracy_score(y_test, y_pred_rf))

# Confusion Matrices
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_dt)
plt.title("Decision Tree Confusion Matrix")
plt.show()

ConfusionMatrixDisplay.from_predictions(y_test, y_pred_rf)
plt.title("Random Forest Confusion Matrix")
plt.show()
```
